```python
import xml.etree.ElementTree as ET
import csv
import os

def xml_to_csv(xml_file, csv_file):
    tree = ET.parse(xml_file)
    root = tree.getroot()
    
    # Trova i campi dalla prima istanza (assumendo una struttura uniforme)
    headers = [elem.tag for elem in root[0]]
    
    with open(csv_file, 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(headers)  # Scrive l'intestazione
        
        for item in root:
            writer.writerow([item.find(h).text if item.find(h) is not None else '' for h in headers])
    
    print(f"Conversione completata! File salvato come: {csv_file}")

```

```python
# Esempio d'uso
xml_filename = "C:\\Users\\lollo\\Downloads\\paolomariapack\\paolomariapack\\Paolo_Verdi.xml"  # Sostituisci con il nome del tuo file XML
csv_filename = os.path.splitext(xml_filename)[0] + ".csv"
xml_to_csv(xml_filename, csv_filename)
```
